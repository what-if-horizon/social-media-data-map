```python
import pandas as pd 
```

```python
root_dir = '/home/bsc/bsc093754/GIT/social-media-data-map/data/processed/'
input_ids = f'{root_dir}annotated_paths.csv'
input_participants = f'{root_dir}participant_data.csv'
output_file = f'{root_dir}participant_data_id_merged.csv'
```

```python
def merge_participant_ids(input_ids, input_participants, output_file):

    df_id = pd.read_csv(input_ids)
    df_part = pd.read_csv(input_participants)

    df_id = df_id[['platform', 'date_added', 'duplicate_flag', 'final_path', 'keepID']]
    df_part = df_part[['participant', 'platform', 'final_path']]
    mon_cols = ['platform','final_path']

    df = pd.merge(df_part, df_id, on=mon_cols, how='inner')
    df = df.drop_duplicates(subset = ['participant', 'platform', 'final_path', 'keepID'])

    df.to_csv(output_file)
    return df

merge_participant_ids(input_ids, input_participants, output_file)
```
