```python
import pandas as pd
import ast
from sentence_transformers import SentenceTransformer
from datetime import datetime, date
```

```python
df = pd.read_csv('/home/bsc/bsc093754/GIT/social-media-data-map/data/processed/path_embeddings_2026-05-18.csv')
model = "/gpfs/projects/bsc100/models/sentence-transformers/all-MiniLM-L6-v2"
```

```python

def compute_sim(model, df):
    df['sim'] = None
    df['path2'] = None
    model = SentenceTransformer(model)
    for ix, row in df.iterrows():
        embeddings1 = ast.literal_eval(row['final_path_emb'])
        sim_list = []
        path2_list = []
        for ixx, row in df.iterrows():
            if ixx == ix:
                continue
            embeddings2 = ast.literal_eval(row['final_path_emb'])
            sim = model.similarity(embeddings1, embeddings2)
            sim = sim.tolist()[0]
            sim_list.append(sim)
            path2_list.append(row['final_path'])
        
        df.at[ix,'sim'] = sim_list
        df.at[ix, 'path2'] = path2_list
    print(df)
    df = df.explode(['sim', 'path2'])
    
    return df


df = compute_sim(model, df)
```
